**Why Embeddings Matters? - Similarity Done**

```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity

model = SentenceTransformer("sentence-transformers/all-MiniLM-L6-v2")

sentences = [
    "I love biryani",
    "Biryani is my favourite food",
    "The car is very fast",
    "Hyderabad is famous for food"
]

embeddings = model.encode(sentences)

query = "I like eating spicy rice food"

query_embedding = model.encode([query])

scores = cosine_similarity(query_embedding, embeddings)[0]

for sentence, score in zip(sentences, scores):
  print(f"{sentence} --> Similarity: {score:.3f}")
```

```python
from transformers import AutoTokenizer
tokenizer = AutoTokenizer.from_pretrained("gpt2")

context_limit = 15

text = """
LangChain is a framework for building applications with large language models.
It helps developers connect prompts, models, memory, tools, documents and output parses.
This makes it easier to build chatbots, RAG systems and AI agents
"""
tokens = tokenizer.encode(text)

print("Total Tokens: ", len(tokens))
print("Context Limit: ", context_limit)

if len(tokens) > context_limit:
  print("Text is too long for the context window")
  allowed_tokens = tokens[: context_limit]
  shortened_text = tokenizer.decode(allowed_tokens)
  print("\n Shortened Text:")
  print(shortened_text)
else:
  print("Text fits inside the context window")
```

```python

import random
import math
from collections import defaultdict, Counter

# --------------------------------
# STEP 1: TRAINING DATA
# --------------------------------

sentences = [
    "I love Python",
    "I love artificial intelligence",
    "I love machine learning",
    "I learn Python",
    "I learn artificial intelligence",
    "Python is a programming language",
    "Python is useful for AI",
    "Machine learning is interesting",
    "Artificial intelligence is powerful",
    "I enjoy learning Python",
    "I enjoy learning AI"
]


# --------------------------------
# STEP 2: TOKENIZATION
# --------------------------------

def tokenize(text):

    return text.lower().split()


# --------------------------------
# STEP 3: CREATE THE MODEL
# --------------------------------

model = defaultdict(Counter)

def train_model(sentences):

    for sentence in sentences:

        tokens = tokenize(sentence)

        for i in range(len(tokens) - 1):

            current_word = tokens[i]

            next_word = tokens[i + 1]

            model[current_word][next_word] += 1

    print("Training completed!")


# --------------------------------
# STEP 4: PREDICT NEXT WORD
# --------------------------------

def predict_next(word):

    predictions = model[word]

    if not predictions:
        return None

    return predictions.most_common(1)[0][0]


# --------------------------------
# STEP 5: TEMPERATURE SAMPLING
# --------------------------------

def generate_next(word, temperature=1.0):

    predictions = model[word]

    if not predictions:
        return None

    words = list(predictions.keys())

    counts = list(predictions.values())

    total = sum(counts)

    probabilities = [
        count / total for count in counts
    ]

    scores = [
        math.log(p) / temperature
        for p in probabilities
    ]

    maximum = max(scores)

    exp_scores = [
        math.exp(x - maximum)
        for x in scores
    ]

    denominator = sum(exp_scores)

    weights = [
        x / denominator for x in exp_scores
    ]

    return random.choices(
        words,
        weights=weights,
        k=1
    )[0]


# --------------------------------
# STEP 6: TEXT GENERATION
# --------------------------------

def generate_text(prompt, length=5, temperature=1.0):

    words = tokenize(prompt)

    for i in range(length):

        next_word = generate_next(
            words[-1],
            temperature
        )

        if next_word is None:
            break

        words.append(next_word)

    return " ".join(words)


# --------------------------------
# STEP 7: TRAIN THE MODEL
# --------------------------------

train_model(sentences)


# --------------------------------
# STEP 8: USER INTERACTION
# --------------------------------

while True:

    print("\n----- MINI LLM -----")

    prompt = input("Enter your text (or exit): ")

    if prompt.lower() == "exit":
        print("Goodbye!")
        break

    temperature = float(
        input("Enter temperature (0.2 to 2.0): ")
    )

    if temperature <= 0:
        print("Temperature must be greater than zero.")
        continue

    answer = generate_text(
        prompt,
        length=5,
        temperature=temperature
    )

    print("\nGenerated Text:")
    print(answer)
```
